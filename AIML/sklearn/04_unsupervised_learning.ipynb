```python
import pandas as pd    
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
data = {
    'Customer':['riya','aman','faizan','neha','imran','sneha'],
    'Age':[20,30,40,22,38,25],
    'Spending':[100,200,300,110,290,130]
}
df = pd.DataFrame(data)
X = df[['Age','Spending']]
#random state means random random_state is used to control randomness in the algorithm so that you get the same result every time you run the code.
#n_init used for checking clutering points 10 times with the data
#random stste is given to reduce the randomness of result
model = KMeans(n_clusters=2,random_state=42,n_init=10)
df['Group']=model.fit_predict(X)

#width of graph is 6 and height is 5
plt.figure(figsize=(6,5))
for group in df['Group'].unique(): #unique gives didtinct values 
    group_data = df[df['Group']==group]#this line is used formasking 
    plt.scatter(group_data['Age'],group_data['Spending'],label=f'Group{group}')
plt.xlabel('Age')
plt.ylabel('Soending score')
plt.title('Customer segmens(K-means)')
plt.legend()
plt.grid(True)
plt.grid(True)
plt.show()


print(df)

```

```python
import pandas as pd
import numpy as np          
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
data = {
    'Income':[30000,40000,50000,60000,70000,80000],
    'Age':[25,30,35,40,45,50],
    'Spending':[70,60,50,40,30,20],
    'Savings':[1000,5000,8000,10000,15000,20000]
    
}

df = pd.DataFrame(data)
#standard scaler is used to standrdise the data in a standard form 
scaler = StandardScaler()
#fit in fit transform used for learing from the standard data 
#transfrom is used to apply the fit login to scale your data
scaled_data = scaler.fit_transform(df)
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(pca_result,columns=['PCA1','PCA2'])
exaplined_variance = pca.explained_variance_ratio_
print("variance cptured by each component: ")
print(np.round(exaplined_variance*100,2))#by giveng round and 2 we are giving answert till 2 decimalplaces 
plt.figure(figsize=(8,6))
plt.scatter(pca_df['PCA1'],pca_df['PCA2'],color='black')
plt.title("PCA projection (2D view)")
plt.xlabel('PCA1 main pattern')
plt.ylabel('PCA2 minor pattern ')
plt.grid(True)
plt.show()

print("New data with two features PCA1 and PCA2")
print(pca_df)
""""   



"""

```
