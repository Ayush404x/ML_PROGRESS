```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
import matplotlib.pyplot as plt   
import numpy as np  


#load data set
data = pd.read_csv("/home/ayush/Documents/ML_Progress/AIML/sklearn/Students_Performance_Dataset_150Rows.csv")


#input

X = data[['Study_Hours_per_Week']]

y = data[['Final_Score']]

#train the model
model = LinearRegression()
model.fit(X,y)
predicted_score = model.predict(X)

#valid regression matrix
mse = mean_squared_error(y,predicted_score)
mae = mean_absolute_error(y,predicted_score)
rmse = np.sqrt(mse)
r2 = r2_score(y,predicted_score)
#show result

"""
if we get value 
3.12345
round(3.12345) 

"""
print("MEAN SQUARE ERROR (MSE) : ",round(mse,2))
print("MEAN ABSOLUTE ERROR (MAE) : ",round(mae,2))
print("ROOT MEAN SQUARE ERROR (RMSE) : ",round(rmse,2))
print("R^2 SCORE (MODEL ACCUARACY) : ",round(r2,2))#if it is closer to 1 then t is better

#drawing histogram and scatter plot 
#(width,height)
plt.figure(figsize=(10,6))
plt.hist(data['Final_Score'],bins = 30,color = 'pink',edgecolor='black')
plt.title("Ditribution of final exam scores")
plt.Xlabel=("Final_score")
plt.ylable=("No_of_students")
plt.grid(True)
plt.show()

#r^2 score is 0.01 means it is givng the avergae value


#plotting regression line using scatter plot
plt.figure(figsize=(10,6))
plt.scatter(X,y,color = 'pink',label='Actual scores')
plt.plot(X,predicted_score,color="blue",label="predicted scores(Regression line)")
plt.title("model prediction and actual scores")
plt.Xlabel=("Study_hours_per_week")
plt.ylable=("Final output")
plt.grid(True)
plt.show()

new_hour = 9
predicted_score_new = model.predict([[new_hour]])
print(f"predicted final score for {new_hour} Hours is {predicted_score_new} Score")

```
