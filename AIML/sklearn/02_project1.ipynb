```python
import numpy as np        
import pandas as pd
from sklearn.metrics import mean_absolute_error,mean_squared_error
from sklearn.linear_model import LinearRegression
data = pd.read_csv("/home/ayush/Documents/ML_Progress/AIML/sklearn/student.csv")

X = data[['Hours']]#double brackets for 2d input
y = data[['Score']]#it is our target column
model = LinearRegression()
model.fit(X,y)

#Evaluate
predicted_score = model.predict(X)
mse = mean_squared_error(y,predicted_score)
mae = mean_absolute_error(y,predicted_score)
rmse = np.sqrt(mse)
 #show the result 
print("Mean absolute Error (MAE) : ",mae)
print("Mean sqaure Error (MSE) : ",mse)
print("Root mean absolute error (RMSE) : ",rmse)


new_hour = float(input("Enter a hour= "))
new_pred = model.predict([[new_hour]])
print(f"prediction for new hour {new_hour} is score = {new_pred}")
```
